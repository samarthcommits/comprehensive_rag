```python
from pymilvus import Collection, connections
def check_collection_size(collection_name = '', username = ''):
    try:
        connections.connect(host="127.0.0.1", port=19530, db_name=username)

        collection = Collection(collection_name)
        collection.load()

        # Query for count
        result = collection.query(
            expr="",  # Empty expression matches all
            output_fields=["count(*)"]
        )
        # json.loads(r/)
        return int(result[0]['count(*)'])
    except Exception as e:
        print(e)
        return 0
    
check_collection_size(collection_name='transformers_dense', username='sam')
```
